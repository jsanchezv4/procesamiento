## Scrapear Enlaces de Artículos

### Subtask:
Desarrollar un script en Python para visitar las páginas de Yahoo Finance, identificar los enlaces a los artículos de noticias y extraer sus URLs. Se debe considerar la paginación si es necesario para obtener un volumen adecuado de noticias.


**Reasoning**:
I need to import the necessary libraries, define the base URL for Yahoo Finance news, and then iterate through a few pages to scrape article links. For each page, I will make an HTTP request, parse the HTML, identify article links, and store their full URLs in a list. Finally, I will print the total number of collected URLs and display a few examples.



````python
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin
import time
import random

# Lista para almacenar las URLs de los artículos
article_urls = []

# URL base de Yahoo Finance para noticias más recientes
base_url = 'https://finance.yahoo.com/topic/latest-news/'

# Parámetros de paginación
pages_to_scrape = 2
articles_per_page = 25 # Number of articles per page, typical for Yahoo Finance

# Define a User-Agent header
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
}

print(f"Iniciando el scraping de {pages_to_scrape} páginas de noticias de Yahoo Finance desde: {base_url}")

for i in range(pages_to_scrape):
    offset = i * articles_per_page
    paged_url = f"{base_url}?offset={offset}&count={articles_per_page}"

    print(f"\nScrapeando página: {paged_url}")

    try:
        response = requests.get(paged_url, headers=headers, timeout=10)
        response.raise_for_status()

        soup = BeautifulSoup(response.text, 'html.parser')
        links = soup.find_all('a', href=True)

        found_on_page = 0
        for link in links:
            href = link['href']
            full_url = urljoin(paged_url, href)

            # Refined filtering for Yahoo Finance news articles
            # Check if the full_url starts with the specific finance news path
            # and ends with .html, while excluding known non-article patterns.
            if (full_url.startswith('https://finance.yahoo.com/news/') and
                full_url.endswith('.html') and
                '/news/archive/' not in full_url and
                '/m/news/' not in full_url):
                if full_url not in article_urls:
                    article_urls.append(full_url)
                    found_on_page += 1
        print(f"Se encontraron {found_on_page} nuevos enlaces en esta página.")

    except requests.exceptions.RequestException as e:
        print(f"Error al scrapear {paged_url}: {e}")

    sleep_time = random.uniform(2, 5)
    print(f"Esperando {sleep_time:.2f} segundos antes de la siguiente solicitud...")
    time.sleep(sleep_time)

print(f"\nSe han recolectado un total de {len(article_urls)} URLs de artículos.")
print("Primeras 5 URLs de artículos recolectadas:")
for url in article_urls[:5]:
    print(url)
````

### Subtask:
Iterar a través de la lista de URLs de artículos, realizar solicitudes HTTP para cada URL, analizar el contenido HTML con BeautifulSoup y extraer el título, la fecha de publicación y el texto completo de cada artículo.


**Reasoning**:
I need to implement the logic for extracting article content (title, date, text) from the collected URLs. This involves iterating through `article_urls`, making requests, parsing HTML with BeautifulSoup, and handling potential errors and delays as per the instructions.



````python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time
import random
from datetime import datetime
from dateutil import parser

articles_data = []
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
}

print(f"Iniciando la extracción de contenido para {len(article_urls)} artículos.")

for i, url in enumerate(article_urls):
    print(f"\nProcesando artículo {i+1}/{len(article_urls)}: {url}")
    title = 'N/A'
    publication_date = 'N/A'
    text_content = 'N/A'

    try:
        response = requests.get(url, headers=headers, timeout=15)
        response.raise_for_status()

        soup = BeautifulSoup(response.text, 'html.parser')

        # Debug: Check if a known paragraph class is found in the raw soup
        if soup.find('p', class_='yf-1fy9kyt'):
            print("  -> DEBUG: 'yf-1fy9kyt' class found in the raw HTML fetched by requests.")
        else:
            print("  -> DEBUG: 'yf-1fy9kyt' class NOT found in the raw HTML fetched by requests.")
        if soup.find('div', class_='caas-body'):
            print("  -> DEBUG: 'caas-body' container found in the raw HTML fetched by requests.")
        else:
            print("  -> DEBUG: 'caas-body' container NOT found in the raw HTML fetched by requests.")

        # --- Extract Title ---
        h1_title_element = soup.find('header', class_='caas-header')
        if h1_title_element:
            h1_tag = h1_title_element.find('h1')
            if h1_tag:
                title = h1_tag.get_text(strip=True)

        if title == 'N/A':
            meta_title = soup.find('meta', property='og:title')
            if meta_title and meta_title.get('content'):
                title = meta_title['content'].strip()

        if title == 'N/A' and soup.title:
            title = soup.title.get_text(strip=True)


        # --- Extract Publication Date ---
        date_str = None
        time_tag = soup.find('time', datetime=True)
        if time_tag:
            date_str = time_tag['datetime'].strip()
        else:
            meta_date = soup.find('meta', property='article:published_time')
            if meta_date and meta_date.get('content'):
                date_str = meta_date['content'].strip()
            else:
                span_date = soup.find('span', {'data-testid': 'article-timestamp'})
                if span_date:
                    date_str = span_date.get_text(strip=True)
                else:
                    div_time = soup.find('div', class_='caas-attr-time-style')
                    if div_time:
                        time_in_div = div_time.find('time')
                        if time_in_div and time_in_div.get('datetime'):
                            date_str = time_in_div['datetime'].strip()

        if date_str:
            try:
                dt_obj = parser.parse(date_str)
                publication_date = dt_obj.strftime('%Y-%m-%d %H:%M:%S')
            except Exception as e:
                print(f"Error al parsear la fecha '{date_str}' con dateutil: {e}")
                publication_date = date_str


        # --- Extract Article Text ---
        extracted_text = ""
        # Main content area selectors, ordered by specificity/reliability
        content_areas = [
            'div[data-test-id="Main_Stream_Article_Content"]',
            'div[data-test-id="article-body"]', # Very common for Yahoo Finance
            'div.caas-body', # Also very common
            'article[itemprop="articleBody"]', # Semantic HTML
            'div[itemprop="articleBody"]',
            'div#editor-container', # Older Yahoo articles
            'div.canvas-body',
            'div.story-body',
            'div.article-content', # More generic
        ]

        # Elements or classes to typically remove if found within the article body
        unwanted_elements_in_body = [
            'script', 'style', 'aside', 'nav', 'footer', 'header', # Structural
            '.caas-sdk-asset', '.caas-text-byline', '.caas-content-byline',
            '.caas-infobox', '.caas-footnote', '.caas-article-snippet',
            '.embed-container', '.caas-related', '.caas-read-more',
            '.caas-summary-wrapper', '.caas-stream-top-stories-cards',
            '.caas-sub-wrapper', '.caas-social-share-bar', '.caas-ad',
            '.caas-trending-container', 'div[data-component="LinkableText"]',
            'div[class*="social"]', 'div[class*="share"]',
            '.read-more-button', '.embed-youtube', '.player-service-wrapper',
            'div.comments-nav', 'div.comments-container', # Comments sections
            'div[id*="ad-"]', 'div[class*="ad-"]', # Generic ad indicators
            'div[class*="promo"]', # Promotional content
            'div[class*="disclaimer"]', # Disclaimers
            'p[class*="credit"]', 'p[class*="caption"]' # Image credits/captions
        ]

        for selector in content_areas:
            container = soup.select_one(selector)
            if container:
                print(f"  -> Attempting to extract text from container: {selector}")
                # Create a temporary soup for this container to safely remove elements
                temp_container_soup = BeautifulSoup(str(container), 'html.parser')

                # Remove unwanted elements first
                for unwanted_sel in unwanted_elements_in_body:
                    for tag in temp_container_soup.select(unwanted_sel):
                        tag.decompose()

                # Extract all significant text elements (paragraphs, list items)
                content_elements = temp_container_soup.find_all(['p', 'li', 'div']) # Include divs for direct text

                # Filter out very short or empty elements that might be noise
                valid_texts = []
                for elem in content_elements:
                    text = elem.get_text(strip=True)
                    if len(text) > 30: # Heuristic: text must be reasonably long
                        valid_texts.append(text)

                extracted_text = '\n\n'.join(valid_texts)

                if extracted_text.strip():
                    print(f"  -> Successfully extracted text using selector '{selector}'. Sample: {extracted_text[:300]}...")
                    break # Found content, stop trying other selectors
                else:
                    print(f"  -> No significant text extracted from container with selector '{selector}'.")
            else:
                print(f"  -> Selector '{selector}' did not find any container.")

        # Fallback if no specific content area yielded text
        if not extracted_text.strip():
            print(f"  -> WARNING: No specific article content container found or extracted text was empty. Attempting global paragraph search.")
            all_paragraphs = soup.find_all('p')
            valid_paragraphs = []
            for p in all_paragraphs:
                text = p.get_text(strip=True)
                # Avoid very short paragraphs, common navigational elements, etc.
                if len(text) > 50 and not any(kw in text.lower() for kw in ['read more', 'continue reading', 'subscribe', 'comments']):
                    valid_paragraphs.append(text)
            extracted_text = '\n\n'.join(valid_paragraphs)
            if extracted_text.strip():
                print(f"  -> GLOBAL FALLBACK: Extracted text from all paragraphs on the page. Sample: {extracted_text[:150]}...")
            else:
                print(f"  -> GLOBAL FALLBACK: Found paragraphs, but no significant text after filtering.")

        if not extracted_text.strip():
            text_content = "No se pudo extraer el texto del artículo."
            print(f"  -> Final result: Failed to extract any article text for {url}")
        else:
            cleaned_lines = []
            for line in extracted_text.splitlines():
                stripped_line = line.strip()
                if stripped_line:
                    cleaned_lines.append(stripped_line)
            text_content = '\n\n'.join(cleaned_lines)
            print(f"  -> Final result: Successfully extracted text for {url}. Total length: {len(text_content)}")

    except requests.exceptions.RequestException as e:
        print(f"Error de red al scrapear {url}: {e}")
        title = f"ERROR: Network issue - {e}"
        text_content = f"ERROR: Could not retrieve article due to network issue: {e}"
        publication_date = "N/A"
    except Exception as e:
        print(f"Error general al procesar {url}: {e}")
        title = f"ERROR: General issue - {e}"
        text_content = f"ERROR: Could not process article due to: {e}"
        publication_date = "N/A"

    articles_data.append({
        'url': url,
        'title': title,
        'publication_date': publication_date,
        'text': text_content
    })

    sleep_time = random.uniform(1, 3)
    print(f"Esperando {sleep_time:.2f} segundos antes de la siguiente solicitud...")
    time.sleep(sleep_time)

print(f"\nExtracción de contenido finalizada. Se procesaron {len(articles_data)} artículos.")

df_articles = pd.DataFrame(articles_data)
print("Primeras 5 filas del DataFrame de artículos:")
print(df_articles.head())
print("\nInformación del DataFrame:")
df_articles.info()
print("\nEjemplo de texto extraído (primer artículo):")
if not df_articles.empty:
    print(df_articles.loc[0, 'text'][:500] + "...")
````

````python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time
import random
from datetime import datetime
from dateutil import parser
import asyncio
from playwright.async_api import async_playwright

# Lista para almacenar los datos extraídos de los artículos
articles_data = []

# Define a User-Agent header (might still be useful for initial requests or if Playwright needs it explicitly)
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4473.85 Safari/537.36'
}

async def scrape_articles_with_playwright(article_urls):
    global articles_data # Declare articles_data as global to modify the list outside the function
    print(f"Iniciando la extracción de contenido para {len(article_urls)} artículos usando Playwright.")

    async with async_playwright() as p:
        browser = await p.chromium.launch(headless=True)
        page = await browser.new_page()

        for i, url in enumerate(article_urls):
            print(f"\nProcesando artículo {i+1}/{len(article_urls)}: {url}")
            title = 'N/A'
            publication_date = 'N/A'
            text_content = 'N/A'

            try:
                await page.goto(url, timeout=60000) # Navigate to the URL, 60 seconds timeout
                # Wait for some common element that indicates the article content is loaded
                await page.wait_for_selector('div.caas-body, p.yf-1fy9kyt, div[data-test-id="article-body"]', timeout=30000)

                # Get the fully rendered HTML content
                html_content = await page.content()
                soup = BeautifulSoup(html_content, 'html.parser')

                # Debug: Check if a known paragraph class is found in the rendered soup
                if soup.find('p', class_='yf-1fy9kyt'):
                    print("  -> DEBUG: 'yf-1fy9kyt' class found in the RENDERED HTML.")
                else:
                    print("  -> DEBUG: 'yf-1fy9kyt' class NOT found in the RENDERED HTML.")
                if soup.find('div', class_='caas-body'):
                    print("  -> DEBUG: 'caas-body' container found in the RENDERED HTML.")
                else:
                    print("  -> DEBUG: 'caas-body' container NOT found in the RENDERED HTML.")

                # --- Extract Title ---
                h1_title_element = soup.find('header', class_='caas-header')
                if h1_title_element:
                    h1_tag = h1_title_element.find('h1')
                    if h1_tag:
                        title = h1_tag.get_text(strip=True)

                if title == 'N/A':
                    meta_title = soup.find('meta', property='og:title')
                    if meta_title and meta_title.get('content'):
                        title = meta_title['content'].strip()

                if title == 'N/A' and soup.title:
                    title = soup.title.get_text(strip=True)


                # --- Extract Publication Date ---
                date_str = None
                time_tag = soup.find('time', datetime=True)
                if time_tag:
                    date_str = time_tag['datetime'].strip()
                else:
                    meta_date = soup.find('meta', property='article:published_time')
                    if meta_date and meta_date.get('content'):
                        date_str = meta_date['content'].strip()
                    else:
                        span_date = soup.find('span', {'data-testid': 'article-timestamp'})
                        if span_date:
                            date_str = span_date.get_text(strip=True)
                        else:
                            div_time = soup.find('div', class_='caas-attr-time-style')
                            if div_time:
                                time_in_div = div_time.find('time')
                                if time_in_div and time_in_div.get('datetime'):
                                    date_str = time_in_div['datetime'].strip()

                if date_str:
                    try:
                        dt_obj = parser.parse(date_str)
                        publication_date = dt_obj.strftime('%Y-%m-%d %H:%M:%S')
                    except Exception as e:
                        print(f"Error al parsear la fecha '{date_str}' con dateutil: {e}")
                        publication_date = date_str


                # --- Extract Article Text ---
                extracted_text = ""
                found_main_content = False

                # Prioritize specific paragraph class identified by user, then main content containers
                main_content_selectors = [
                    'p.yf-1fy9kyt', # Direct hit for user's provided example
                    'div[data-test-id="Main_Stream_Article_Content"]',
                    'div[data-test-id="article-body"]',
                    'div.caas-body',
                    'div.article-uuid',
                    'article[itemprop="articleBody"]',
                    'div[itemprop="articleBody"]',
                    'div#editor-container',
                    'div.article-body',
                    'section.article-content',
                    'div.canvas-body',
                    'div.story-body',
                ]

                for selector in main_content_selectors:
                    if selector.startswith('p.') or selector.startswith('div.yf-'): # Handle direct paragraph selectors
                        elements = soup.select(selector)
                        if elements:
                            print(f"  -> Attempting to extract text directly from elements with selector: {selector}")
                            extracted_text = '\n\n'.join([e.get_text(strip=True) for e in elements if e.get_text(strip=True)])
                            if extracted_text.strip():
                                print(f"  -> Successfully extracted text using selector '{selector}'. Sample: {extracted_text[:300]}...")
                                found_main_content = True
                                break
                    else:
                        container = soup.select_one(selector)
                        if container:
                            print(f"  -> Attempting to extract text from container: {selector}")
                            temp_container_soup = BeautifulSoup(str(container), 'html.parser')

                            # Elements or classes to typically remove if found within the article body
                            unwanted_elements_in_body = [
                                'script', 'style', 'aside', 'nav', 'footer', 'header',
                                '.caas-sdk-asset', '.caas-text-byline', '.caas-content-byline',
                                '.caas-infobox', '.caas-footnote', '.caas-article-snippet',
                                '.embed-container', '.caas-related', '.caas-read-more',
                                '.caas-summary-wrapper', '.caas-stream-top-stories-cards',
                                '.caas-sub-wrapper', '.caas-social-share-bar', '.caas-ad',
                                '.caas-trending-container', 'div[data-component="LinkableText"]',
                                'div[class*="social"]', 'div[class*="share"]',
                                '.read-more-button', '.embed-youtube', '.player-service-wrapper',
                                'div.comments-nav', 'div.comments-container',
                                'div[id*="ad-"]', 'div[class*="ad-"]',
                                'div[class*="promo"]',
                                'div[class*="disclaimer"]',
                                'p[class*="credit"]', 'p[class*="caption"]'
                            ]

                            for unwanted_sel in unwanted_elements_in_body:
                                for tag in temp_container_soup.select(unwanted_sel):
                                    tag.decompose()

                            content_elements = temp_container_soup.find_all(['p', 'li', 'div'])
                            valid_texts = []
                            for elem in content_elements:
                                text = elem.get_text(strip=True)
                                if len(text) > 30:
                                    valid_texts.append(text)
                            extracted_text = '\n\n'.join(valid_texts)

                            if extracted_text.strip():
                                print(f"  -> Successfully extracted text using container selector '{selector}'. Sample: {extracted_text[:300]}...")
                                found_main_content = True
                                break
                            else:
                                print(f"  -> No significant text extracted from container with selector '{selector}'.")

                # Fallback if no specific content area yielded text
                if not found_main_content and not extracted_text.strip():
                    print(f"  -> WARNING: No specific article content container found or extracted text was empty. Attempting global paragraph search.")
                    all_paragraphs = soup.find_all('p')
                    valid_paragraphs = []
                    for p in all_paragraphs:
                        text = p.get_text(strip=True)
                        if len(text) > 50 and not any(kw in text.lower() for kw in ['read more', 'continue reading', 'subscribe', 'comments']):
                            valid_paragraphs.append(text)
                    extracted_text = '\n\n'.join(valid_paragraphs)
                    if extracted_text.strip():
                        print(f"  -> GLOBAL FALLBACK: Extracted text from all paragraphs on the page. Sample: {extracted_text[:150]}...")
                    else:
                        print(f"  -> GLOBAL FALLBACK: Found paragraphs, but no significant text after filtering.")

                if not extracted_text.strip():
                    text_content = "No se pudo extraer el texto del artículo."
                    print(f"  -> Final result: Failed to extract any article text for {url}")
                else:
                    cleaned_lines = []
                    for line in extracted_text.splitlines():
                        stripped_line = line.strip()
                        if stripped_line:
                            cleaned_lines.append(stripped_line)
                    text_content = '\n\n'.join(cleaned_lines)
                    print(f"  -> Final result: Successfully extracted text for {url}. Total length: {len(text_content)}")

            except Exception as e:
                print(f"Error general al procesar {url}: {e}")
                title = f"ERROR: General issue - {e}"
                text_content = f"ERROR: Could not process article due to: {e}"
                publication_date = "N/A"

            articles_data.append({
                'url': url,
                'title': title,
                'publication_date': publication_date,
                'text': text_content
            })

            # Add a random delay between requests
            sleep_time = random.uniform(2, 7) # Increased delay to be more polite/avoid blocks
            print(f"Esperando {sleep_time:.2f} segundos antes de la siguiente solicitud...")
            await asyncio.sleep(sleep_time)

        await browser.close()

    print(f"\nExtracción de contenido finalizada. Se procesaron {len(articles_data)} artículos.")
    return articles_data

# Run the async scraping function
articles_data = asyncio.run(scrape_articles_with_playwright(article_urls))

df_articles = pd.DataFrame(articles_data)
print("Primeras 5 filas del DataFrame de artículos:")
print(df_articles.head())
print("\nInformación del DataFrame:")
df_articles.info()
print("\nEjemplo de texto extraído (primer artículo):")
if not df_articles.empty:
    print(df_articles.loc[0, 'text'][:500] + "...")
````
